```python
# conditional program

marks = int(input("Enter your Maths marks: "))

if marks > 95:
    print("Grade A+")
elif marks >= 75:
    print("Grade A")
elif marks >= 65:
    print("Grade B")
elif marks >=55:
    print("Grade C")
elif marks >=45:
    print("Grade D")    
else:
    print("Grade E")
```

```python
# Nested if
uname = "wscube"
password = "123456"

name = input("Enter user name:")
ps = input("Enter user password:")

if uname == name:
    print("Username is correct")
    if(password==ps):
        print("Welcome back!!")
    else:
        print("Password is incorrect!!")
else:
    print("Incorrect username!!")
```

```python
# Write a program to check the no of vowels inside the given string

name = input("Enter any name to check the no of vowels it containes: ")

count = 0
for c in name:
    if c.lower() in ("a","e","i","o","u"):
        count += 1

print(count)
```

```python
# multiplication table using loops

no = int(input("Enter the no for which you want the multiplication table: "))

for i in range(1, 11, 1):
    print("", no, "x", i, "=", (no*i))
```

```python
# factorial 
# 5! = 5*4*3*2*1

fact = 1
no = int(input("Enter no for factorial:"))

for i in range(1,no+1,1):
    fact *= i
    print(i, fact)
```
