```python
# math
import math

print(math.pow(5,3))
print(math.sqrt(64))
print(math.floor(3.4536))
print(math.ceil(3.4536))
print(math.trunc(3.4536))
print(math.factorial(5))
print(math.pi)
```

```python
# datetime
import datetime as dt

d = dt.datetime.now()
print(d.day)
print(d.month)
print(d.year)
print(d.weekday())
print(d.today())
print(d.date())
```

```python
d = dt.date(2026, 11, 21)
print(d)
```

```python
strDate = "28-10-2026"

d = dt.datetime.strptime(strDate, "%d-%m-%Y")
print(d.strftime("%d-%b-%Y-%A"))
```
