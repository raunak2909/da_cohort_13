```python
import numpy as np
import statistics as stats

a = [800, 300, 500, 700, 400]
b = [6, 7, 3, 1, 4]

c = [
    [2,3],
    [7,8],
    [5,6],
    [3,2]
]

x = np.array(a)
y = np.array(b)
z = np.array(c)

print(x+y)
print(np.sum(z, axis = 1))
print(np.mean(y))
print(np.median(z))
print(stats.mode(a))
print(np.std(x))
print(np.var(x))
print(np.ptp(y))
print(np.corrcoef(x,y))
```

```python
print(np.cumsum(x))
```

```python
print(np.cumprod(y))
```

```python
a = [800, 300, np.nan, 700, np.nan]
b = [6, np.nan, 3, 1, 4]

x = np.array(a)
y = np.array(b)

print(np.nansum(y))
print(np.nanmean(y))
print(np.nanmedian(x))
print(np.nanstd(x))
print(np.nanvar(x))
print(stats.mode(x))
```

```python
# slicing 

a = [
     [
         [3,4,5], 
         [4,6,5], 
         [4,3,2], 
         [3,8,9]
     ],
     [
         [2,5,7],
         [3,6,9],
         [1,4,7],
         [2,9,5]
     ],
     [
         [7,8,9],
         [2,5,4],
         [3,9,7],
         [1,8,2],
     ]
]

x = np.array(a)

print(x[-2:-1])
print(x[-2])
```
