```python
# fetching data from api
import requests as req

url = "https://jsonplaceholder.typicode.com/users"
try:
    request = req.get(url)
    
    request.raise_for_status()
    
    if request.status_code == 200:
        data = request.json()
        df = pd.DataFrame(data)
        df
    else:
        print("Something went wrong")
except:
    print("Something went wrong")
df    
```
