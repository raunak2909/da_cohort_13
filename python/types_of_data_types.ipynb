```python
names = ["organic", "pop", "system", "clinic", "thunder", "gathering"]
# Write a program to identify count of words 
# from the given list which has chars more than 5 
# and thier 1st and last char are same

count = 0

for w in names:
    if len(w) > 5 and w[0]==w[-1]:
        count += 1


print(count)
```

```python
no = [2,3,5,7,43,41,87,65,34,22,9,8]

odd = []
even = []

for eachNo in no:
    if eachNo%2 == 0:
        even.append(eachNo)
    else:
        odd.append(eachNo)

print(even, odd)
```
