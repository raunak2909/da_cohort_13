```python
# loop to add two no again and again

addNumbers = "Yes"

while addNumbers in ("yes", "Yes", "YES", "Y", "y"):
    n1 = int(input("Enter no1:"))
    n2 = int(input("Enter no2:"))

    print(f"The sum of {n1} and {n2} is {n1+n2}")
    addNumbers = input("Do you want to add more numbers")
```

```python
# while true

while True:
    n1 = int(input("Enter no1:"))
    n2 = int(input("Enter no2:"))

    print(f"The sum of {n1} and {n2} is {n1+n2}")
    addNumbers = input("Do you want to add more numbers")
    if addNumbers in ("No", "no", "n", "NO", "N"):
        break
```

```python
# nested loop
# billing system


while True:

    name = input("Enter name:")
    mobNo = input("Enter mobNo:")
    amount = 0
    
    
    while True:
        price = float(input("Enter price of Product: "))
        qty = float(input("Enter the quantity: "))
        
        amount += price*qty
        addMore = input("Do you have more products in cart? ")
        if addMore in ("No", "no", "N", "n", "NO"):
            break
    print("-"*50)
    print("Name: ", name)
    print("MobNo: ", mobNo)
    print("-"*50)
    print("Amount: ", amount)
    print("-"*50)

    moreUser = input("Do you have more customer in queue? ")
    if moreUser in ("No", "no", "N", "n", "NO"):
        break

```
