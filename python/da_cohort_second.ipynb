```python
# Comparison operators - boolean value - boolean
# ==
# !=
# >
# >=
# <
# <=

n1 = 34

print(n1 >= 14)


```

```python
# Logical operators - and, or, not - boolean

n1 = 11
n2 = 16

print(not (n1>10 and n2<15))
```

```python
# Membership operators - in, not in - boolean

print("W" not in "Wscube")
```

```python
# Identity operators - is, is not - boolean
n1 = 11
n2 = 11

n1 = [11,12,13]
n2 = [11,12,13]

print(id(n1))
print(id(n2))

print(n1==n2) # checking values

print(n1 is n2) # checking memory
```

```python
a = [1,2,3]
b = [1,2,3]

print(id(a))
print(id(b))

print(a==b)
print(a is b)
```

```python
# assignment operators - +=, -=. *=..

n1 = 11
n1 = n1 * 20
print(n1)
```

```python
amount = 0

amount += 2000
print(amount)
amount -= 4000
print(amount)
```

```python
product = "Smart Phone"
sellingPrice =  20000
discount = "10"
actualPrice = 15000
qty = 10

sales = sellingPrice * qty
totalDiscount = sales * (float(discount)/100)
netRevenue = sales - totalDiscount

totalCost = actualPrice * qty
profit  = netRevenue - totalCost

profitMargin = (profit//netRevenue) * 100

print(f"""
      Product : {product}
      sellingPrice : {sellingPrice}
      discount : {discount}
      actualPrice : {actualPrice}
      qty : {qty}
      sales : {sales}
      totalDiscount : {totalDiscount}
      netRevenue : {netRevenue}
      totalCost : {totalCost}
      profit : {profit}
      profitMargin : {profitMargin}
""")
```

```python
# Conditional Programming

age = int(input("Enter your age: "))

if age>=18:
  print("You are eligible to vote!!")

else:
  print("You're not eligible to vote!!")
print("bye")
```

```python
# A Program to check if the given no is even or odd

no = int(input("Enter any no to check if it is even or odd: "))

if no%2 == 0:
  print(f"The {no} is an EVEN no")
else:
  print(f"The {no} is an ODD no")
```
