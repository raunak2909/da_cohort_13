```python
# We need to create a movie ticket system using dict and loops

movie_db = {
    "Ramayan" : {
        "price" : 1000,
        "seats" : 20
    },

    "Drishyam" : {
        "price" : 500,
        "seats" : 40
    },

    "Hanuman Ansh" : {
        "price" : 700,
        "seats" : 10
    },

    "Udta Teer" : {
        "price" : 400,
        "seats" : 50
    },
}

# print all the movies

#for names in movie_db.keys():
#    print(names)

# search a movie

#movieName = input("Enter the movie to search: ")

#if movieName in movie_db.keys() :
#    print("Yes, Movie found :", movie_db[movieName])
#else:
#    print("No, movie found!!")

# book a movie ticket
while True:
    print("Movie on the BoxOffice right now: ")
    for names, details in movie_db.items():
        print(names, ":", details)

    choice = int(input("\nEnter \n1 for Booking a Ticket\n2 for Cancelling a Ticket and \n3 for Exit:\n"))
    
    if choice == 1:
        movie = input("Enter the movie name:")
            
        if movie in movie_db.keys() :
            print("Yes, Movie found :", movie_db[movie])
            noOfSeats = int(input("Enter no of seats you want to book: "))
        
            if noOfSeats<=movie_db[movie]["seats"]:
                print("-"*50)
                print(f"{noOfSeats} seats reserved successfully!!")
                print("Bill : ", movie_db[movie]["price"]*noOfSeats)
                movie_db[movie]["seats"] -= noOfSeats
                print("-"*50)
                print(movie_db[movie]["seats"], "available..")
            else:
                print("Insufficient seats or invalid no of seats")
        
        else:
            print("Movie not found..")

    elif choice == 2:
        movie = input("Enter the movie name:")

        if movie in movie_db.keys() :
            print("Yes, Movie found")
            noOfSeats = int(input("Enter no of seats you want to cancel: "))
            print("-"*50)
            print(f"{noOfSeats} seats cancelled successfully with 50% deduction!!")
            print("Amount to be returned : ", (movie_db[movie]["price"]*noOfSeats) * 0.5)
            movie_db[movie]["seats"] += noOfSeats
            print("-"*50)
            print(movie_db[movie]["seats"], "available..")
        
        else:
            print("Movie not found..")

    else:
        print("Thank you!!")
        break
        

```
