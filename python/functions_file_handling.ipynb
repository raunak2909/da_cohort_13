```python
# Functions
# block of code - set of instructions which gets executed only when it is called
# helps us to reduce set of repeating statements - redundancy
# helps us to make our code reusable

def multiply(a=1,b=1):
    print("This is Multiplication function")
    product = a*b
    print(f"The sum of {a} and {b} is {product}")

def greet():
    print("Hello World!!")

greet()
multiply(5,11)
multiply(b=100,a=20)
multiply(34)
multiply(23,11)
multiply()
```

```python
def even_odd(n):
    if n % 2 == 0:
        print(f"{n} is an EVEN no.")
    else:
        print(f"{n} is an ODD no.")

even_odd(int(input("Enter any no to check if it is EVEN or ODD: ")))
```

```python
# print vs return 

def check_age(age):
    if(age>=18):
        return True
        print("Hi' you're adult") 
        
    else:
        return False
        print("Hi' you're child")    

    

# child
# print("""Make a choice:
# Enter 1 to learn Alphbets,
# Enter 2 to play indoor games""")     

value = check_age(int(input("Enter your age: "))) 

if value:
    # adult
    print("""Make a choice:
    Enter 1 to vote,
    Enter 2 to play outdoor games""") 

else :
    # child
    print("""Make a choice:
    Enter 1 to learn Alphbets,
    Enter 2 to play indoor games""")     
```

```python
# user defined functions

def gradeSystem(marks):
    if(marks>90):
        return "A+"
    elif marks>75:
        return "A"
    elif marks>65:
        return "B"
    elif marks>55:
        return "C"
    elif marks>45:
        return "D"
    elif marks>33:
        return "E"
    else:
        return "Fail"

grade = gradeSystem(63) 
print(grade)
```

```python
def remove_duplicates(data):
    return list(set(data))

a = [3,2,1,2,5,6,3,2,1,7,8,5,6,4]
b = remove_duplicates(a)
print(a)
print(b)
```

```python
# recursive functions
# def greet():
#     print("Good Morning!!")
#     greet()

# greet()    

# 5! = 5*4*3*2*1 = 120 -> n * n-1
def fact(n):
    if n==1: # terminating condition
        return 1

    return n * fact(n-1)

# 5 * fact(4)
# 5 * 4 * fact(3)
# 5 * 4 * 3 * fact(2)
# 5 * 4 * 3 * 2 * fact(1)
# 5 * 4 * 3 * 2 * 1
fact(5)    
```

```python
# file handling
# read
# write 
# append

f = open("demo.txt", "w")
f.write("Hi, Good Evening!!")
f.close()
```

```python
f = open("demo.txt", "a")
f.write(2000)
f.close()
```

```python
f = open("demo.txt", "r")
data = f.readlines()
print(data)
f.close()
```

```python
with open("demo.txt", "r") as f:
    data = f.readlines()
    print(data)
```

```python
import pickle as p

f = open("demo.txt", "wb")
p.dump(2000, f)
f.close()
```

```python
with open("demo.txt", "rb") as f:
    data = p.load(f)
    print(data)
```
